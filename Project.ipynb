```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("/content/professional_age_gender_dataset.csv")

# View basic info
print("Initial shape:", df.shape)
print("Column names:", df.columns.tolist())
print(df.head())

# 1. Clean column names (strip whitespace, lower case, replace spaces with underscores)
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# 2. Drop duplicate rows
df.drop_duplicates(inplace=True)
print("After dropping duplicates:", df.shape)

# 3. Handle missing values
missing = df.isnull().sum()
print("\nMissing values per column:\n", missing)

# Fill missing numeric values with mean, categorical with mode
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].mean())

# 4. Convert data types if needed (e.g., 'age' might be stored as string)
# Example: Convert 'age' to numeric if it's not
if 'age' in df.columns:
    df['age'] = pd.to_numeric(df['age'], errors='coerce')

# 5. Optional: Remove outliers using Z-score for numerical columns
from scipy.stats import zscore

numeric_cols = df.select_dtypes(include=np.number).columns
z_scores = np.abs(zscore(df[numeric_cols]))
df = df[(z_scores < 3).all(axis=1)]  # keep rows with z-score < 3

print("After removing outliers:", df.shape)

# 6. Final cleaned dataset preview
print("\nCleaned DataFrame preview:")
print(df.head())

# Save cleaned version if needed
# df.to_csv('/mnt/data/cleaned_dataset.csv', index=False)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("/content/professional_age_gender_dataset.csv")


# Optional: Clean column names for consistency
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# Encode categorical features to numeric (required for correlation/covariance)
df_encoded = df.copy()
for col in df_encoded.select_dtypes(include='object').columns:
    df_encoded[col] = df_encoded[col].astype('category').cat.codes

# 1. Compute correlation matrix
correlation_matrix = df_encoded.corr()

# 2. Compute covariance matrix
covariance_matrix = df_encoded.cov()

# Print correlation & covariance
print("Correlation matrix:\n", correlation_matrix)
print("\nCovariance matrix:\n", covariance_matrix)

# 3. Visualize correlation matrix
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# 4. Visualize covariance matrix
plt.figure(figsize=(10, 8))
sns.heatmap(covariance_matrix, annot=True, cmap='viridis', fmt='.2f')
plt.title('Covariance Matrix')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Load and encode dataset
df = pd.read_csv("/content/professional_age_gender_dataset.csv")
df_encoded = df.copy()
for col in df_encoded.select_dtypes(include='object').columns:
    df_encoded[col] = df_encoded[col].astype('category').cat.codes

# Set features and target (assume 'age' is target)
X = df_encoded.drop('age', axis=1)
y = df_encoded['age']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Train linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Equation
print("Linear Regression Equation:")
print("y = {:.2f}".format(model.intercept_), end=' ')
for i, col in enumerate(X.columns):
    print("+ {:.2f}*{}".format(model.coef_[i], col), end=' ')

# Predict and plot (1 feature only for graph)
# Let's use the first feature just for 2D visualization
feature = X.columns[0]
plt.scatter(X_test[feature], y_test, color='blue', label='Actual')
plt.plot(X_test[feature], model.predict(X_test), color='red', label='Predicted')
plt.xlabel(feature)
plt.ylabel('Age')
plt.title(f'Linear Regression: {feature} vs Age')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load and encode dataset
df = pd.read_csv("/content/professional_age_gender_dataset.csv")
for col in df.select_dtypes(include='object'):
    df[col] = df[col].astype('category').cat.codes

# Set features and target (change 'gender' if needed)
X = df.drop('gender', axis=1)
y = df['gender']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Train XGBoost model
model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")

# Plot feature importance
plt.figure(figsize=(10, 6))
plt.barh(X.columns, model.feature_importances_, color='skyblue')
plt.xlabel('Feature Importance')
plt.title('XGBoost Feature Importance')
plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import Lasso, Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Load and encode dataset
df = pd.read_csv("/content/professional_age_gender_dataset.csv")
for col in df.select_dtypes(include='object'):
    df[col] = df[col].astype('category').cat.codes

# Features and target
X = df.drop('age', axis=1)
y = df['age']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Lasso
lasso = Lasso(alpha=1.0)
lasso.fit(X_train, y_train)
y_pred_lasso = lasso.predict(X_test)

# Ridge
ridge = Ridge(alpha=1.0)
ridge.fit(X_train, y_train)
y_pred_ridge = ridge.predict(X_test)

# Metrics
rmse_lasso = np.sqrt(mean_squared_error(y_test, y_pred_lasso))
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))

print("Lasso Regression:")
print(f"RMSE: {rmse_lasso:.2f}, R²: {r2_score(y_test, y_pred_lasso):.2f}")

print("\nRidge Regression:")
print(f"RMSE: {rmse_ridge:.2f}, R²: {r2_score(y_test, y_pred_ridge):.2f}")
```

```python
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Load the dataset
df = pd.read_csv("/content/professional_age_gender_dataset.csv")

# Show column names to identify the correct label column
print("Available columns:", df.columns.tolist())

# ✅ Replace 'Gender' with the correct column name after checking
# For example, assume the actual column name is 'gender' (case sensitive!)
target_column = 'gender'  # <-- Replace this with the actual target name from above list

# Features and target
X = df.drop(target_column, axis=1)
y = df[target_column]

# Encode target labels
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Convert categorical features to numerical
X = pd.get_dummies(X)

# Define ML pipeline: scaling + classifier
model = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression())
])

# Use Stratified K-Folds for classification problems
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Run cross-validation
scores = cross_val_score(model, X, y_encoded, cv=skf, scoring='accuracy')

# Results
print("\nCross-validation scores:", scores)
print("Average accuracy:", scores.mean())
```

```python
import pandas as pd
import numpy as np

# Load dataset

df = pd.read_csv("/content/professional_age_gender_dataset.csv")
# Show column names so you can pick the correct target
print("Columns:", df.columns.tolist())

# ✅ Set your target column name here (adjust if different)
target_column = 'gender'  # <-- change this if needed

# Split features and target
X = df.drop(target_column, axis=1)
y = df[target_column]

# Encode target (e.g., Male/Female → 1/0)
y = y.astype(str)  # make sure it's string
y = (y == 'Male').astype(int).values.reshape(-1, 1)  # Adjust if labels are different

# One-hot encode categorical features
X = pd.get_dummies(X)

# Normalize features (z-score)
X = (X - X.mean()) / X.std()

# Add intercept (bias) term
X = np.c_[np.ones(X.shape[0]), X.values]

# Initialize weights
theta = np.zeros((X.shape[1], 1))

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Gradient Descent Parameters
learning_rate = 0.1
epochs = 1000

# Gradient Descent Loop
for epoch in range(epochs):
    z = X @ theta
    h = sigmoid(z)
    gradient = X.T @ (h - y) / len(y)
    theta -= learning_rate * gradient

    if epoch % 100 == 0:
        loss = -np.mean(y * np.log(h + 1e-5) + (1 - y) * np.log(1 - h + 1e-5))
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

# Final weights
print("\nFinal weights:", theta.ravel())
```
